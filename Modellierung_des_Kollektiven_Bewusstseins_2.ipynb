```python
import numpy as np
import matplotlib.pyplot as plt

# Historische Parameter und Übergangsregeln
EPOCH_PARAMETERS = {
    "Paleolithic": {"tech": 0.05, "population": 0.01, "info_speed": 0.01, "complexity": 0.3},
    "Neolithic": {"tech": 0.15, "population": 0.5, "info_speed": 0.1, "complexity": 0.5},
    "Industrial Age": {"tech": 0.6, "population": 25, "info_speed": 50, "complexity": 0.8},
    "Digital Age": {"tech": 0.95, "population": 150, "info_speed": 299792, "complexity": 0.95}
}

TRANSITION_RULES = {
    "Paleolithic": {"tech_limit": 0.3, "resistance_limit": 0.8, "next": "Neolithic"},
    "Neolithic": {"tech_limit": 0.6, "resistance_limit": 0.7, "next": "Industrial Age"},
    "Industrial Age": {"tech_limit": 0.85, "resistance_limit": 0.6, "next": "Digital Age"}
}

class CivilizationSimulator:
    def __init__(self):
        self.epoch = "Paleolithic"
        self.tech_level = 0.05
        self.social_cohesion = 0.5
        self.history = []

    def calculate_resistance(self):
        params = EPOCH_PARAMETERS[self.epoch]
        return max(0, params["complexity"] - (0.3 * self.social_cohesion))

    def check_transition(self):
        resistance = self.calculate_resistance()
        rule = TRANSITION_RULES.get(self.epoch, {})

        if (self.tech_level > rule.get("tech_limit", 1) and
            resistance < rule.get("resistance_limit", 1)):
            return rule.get("next", self.epoch)
        return self.epoch

    def simulate_step(self):
        # Technologische Entwicklung
        self.tech_level += np.clip(np.random.normal(0.02, 0.01), 0.005, 0.03)

        # Sozialer Wandel
        resistance = self.calculate_resistance()
        self.social_cohesion += (0.5 - resistance) * 0.1

        # Epochenübergang
        new_epoch = self.check_transition()
        if new_epoch != self.epoch:
            print(f"Transition: {self.epoch} → {new_epoch}")
            self.epoch = new_epoch
            self.tech_level *= 1.2  # Innovationsbeschleunigung

        self.history.append({
            "epoch": self.epoch,
            "tech": self.tech_level,
            "resistance": resistance,
            "cohesion": self.social_cohesion
        })

# Simulation ausführen
sim = CivilizationSimulator()
for _ in range(100):
    sim.simulate_step()

# Visualisierung
plt.figure(figsize=(12,6))
plt.plot([x['tech'] for x in sim.history], label='Technologie')
plt.plot([x['resistance'] for x in sim.history], label='Widerstand')
plt.plot([x['cohesion'] for x in sim.history], label='Soziale Kohäsion')
plt.xlabel('Generationen')
plt.ylabel('Intensität')
plt.title('Entwicklung des Kollektiven Bewusstseins')
plt.legend()
plt.show()
```

```python
import numpy as np
from dataclasses import dataclass
from typing import Dict, List
from scipy.special import expit

@dataclass
class NeuroEnhancedState:
    happiness: float = 0.7
    productivity: float = 0.6
    social_flow: float = 0.4
    neurotransmitter_level: float = 1.0
    reuptake_inhibition: float = 0.0
    signal_persistence: float = 1.0
    energy: float = 0.8

    def apply_neuro_modulation(self):
        """Simuliert Amphetaminwirkung durch nichtlineare Transformation"""
        boost = expit(self.neurotransmitter_level * 2 - 1)
        return {
            'happiness': self.happiness * (1 + boost * 0.4),
            'productivity': self.productivity * (1 + boost * 0.6),
            'social_flow': self.social_flow * (1 + self.reuptake_inhibition * 0.3),
            'signal_duration': 1 + self.signal_persistence * 2
        }

class ConsciousnessSimulator:
    def __init__(self):
        self.state = NeuroEnhancedState()
        self.actions = {
            'deep_work': {'base_effect': 0.3, 'neuro_sensitivity': 0.5},
            'social_sync': {'base_effect': 0.2, 'neuro_sensitivity': 0.8}
        }
        self.neuro_history = []

    def simulate_neuro_effect(self, action: str, days: int = 30):
        effect = self.actions[action]
        modulated_effect = effect['base_effect'] * (1 + self.state.neurotransmitter_level * effect['neuro_sensitivity'])

        results = []
        for day in range(days):
            # Signalverstärkung mit Sättigungseffekt
            signal_strength = modulated_effect * (1 - np.exp(-day/7))

            # Neurotransmitter-Dynamik
            self.state.neurotransmitter_level += 0.1 * (1 - self.state.reuptake_inhibition)
            self.state.reuptake_inhibition = min(0.9, self.state.reuptake_inhibition + 0.02)

            # Energiebilanz
            self.state.energy -= 0.05 * signal_strength

            results.append({
                'day': day + 1,
                'productivity': self.state.productivity * (1 + signal_strength),
                'happiness': self.state.happiness * (1 + signal_strength * 0.7),
                'neuro_transmitters': self.state.neurotransmitter_level,
                'signal_power': signal_strength
            })

            # Homöostase-Mechanismus
            if day % 7 == 0:
                self._apply_homeostasis()

        return results

    def _apply_homeostasis(self):
        """Biologischer Regulierungsmechanismus"""
        if self.state.neurotransmitter_level > 2.0:
            self.state.reuptake_inhibition -= 0.1
            self.state.signal_persistence *= 0.9
            print("Neuroadaptation: Rezeptor-Downregulation aktiviert")

    def visualize_neuro_effects(self, results):
        import matplotlib.pyplot as plt

        days = [r['day'] for r in results]
        fig, axs = plt.subplots(2, 2, figsize=(12, 8))

        axs[0,0].plot(days, [r['productivity'] for r in results], 'g-')
        axs[0,0].set_title('Produktivität')

        axs[0,1].plot(days, [r['happiness'] for r in results], 'y-')
        axs[0,1].set_title('Glücksempfinden')

        axs[1,0].plot(days, [r['neuro_transmitters'] for r in results], 'r-')
        axs[1,0].set_title('Neurotransmitter-Level')

        axs[1,1].plot(days, [r['signal_power'] for r in results], 'b-')
        axs[1,1].set_title('Signalstärke')

        plt.tight_layout()
        plt.show()

# Beispiel-Simulation
sim = ConsciousnessSimulator()
results = sim.simulate_neuro_effect('deep_work', days=90)
sim.visualize_neuro_effects(results)
```

```python
# -*- coding: utf-8 -*- 04.11.2026 Andreas Schneider Darmstadt aus Groak Chat: https://grok.com/share/c2hhcmQtMg_acd36f0d-05a6-4458-9f3b-0ad3c20f77c4
"""
Collective Consciousness Simulator – Version 0.11 (Cosmos Forge / Matrixgame)
==============================================================================
Author context : Andreas Schneider + collaborative AI development
Date           : 2026-11-04
Status         : MODEL / GAME  (not empirical science)

Purpose
-------
A modular, regenerable, feedback-capable simulator that integrates:

- Individual state with approximate neurotransmitter dynamics
- Hawkins-inspired consciousness frequency (H) and entropy
- Happiness & Productivity tracking
- Didactic Loop (state → recommended actions → forecast → feedback)
- Simple curriculum / learning-order awareness
- Homeostasis and adaptation mechanisms
- Clear separation of MODEL vs. GAME claims

This script is intentionally practical for personal daily/weekly use and
extensible toward group and collective levels later.

Run:
    python CollectiveConsciousnessSimulator_v011.py

Dependencies: numpy, matplotlib (optional for plots)
"""

from __future__ import annotations
import numpy as np
from dataclasses import dataclass, field
from typing import Dict, List, Tuple, Optional
from enum import Enum
import json
from datetime import datetime
import os

# Optional plotting
try:
    import matplotlib.pyplot as plt
    HAS_MPL = True
except ImportError:
    HAS_MPL = False


# ---------------------------------------------------------------------------
# 1. Core Enums and Constants
# ---------------------------------------------------------------------------

class Mode(Enum):
    REACTIVE = "reactive"      # roughly H < 200
    GUARDIAN = "guardian"      # H >= 200, pro-active / constructive


class ActionType(Enum):
    DEEP_WORK = "deep_work"
    SOCIAL_SYNC = "social_sync"
    REGENERATION = "regeneration"
    CURRICULUM_REVIEW = "curriculum_review"
    SMALL_WIN = "small_win"
    REFLECTION = "reflection"
    PHYSICAL_MOVE = "physical_move"
    LEARNING_SESSION = "learning_session"


# Approximate mapping of Hawkins ranges (MODEL)
HAWKINS_BANDS = {
    "shame_guilt": (1, 50),
    "apathy_grief": (50, 100),
    "fear_desire": (100, 200),
    "anger_pride": (150, 200),
    "courage": (200, 250),          # critical threshold
    "neutrality_willingness": (250, 350),
    "acceptance_reason": (350, 500),
    "love_joy_peace": (500, 700),
    "higher": (700, 1000),
}


# ---------------------------------------------------------------------------
# 2. Neuro-inspired State
# ---------------------------------------------------------------------------

@dataclass
class NeuroState:
    """
    Simplified neurotransmitter / neuromodulator approximation.
    All values are normalized roughly to [0, 2] with 1.0 = baseline.
    This is a MODEL for didactic and motivational purposes only.
    """
    dopamine: float = 1.0          # reward prediction, motivation, learning rate
    serotonin: float = 1.0         # mood stability, impulse control, patience
    norepinephrine: float = 1.0    # arousal, attention, salience
    acetylcholine: float = 1.0     # focus, plasticity gating, learning readiness

    def clip(self):
        self.dopamine = float(np.clip(self.dopamine, 0.1, 2.5))
        self.serotonin = float(np.clip(self.serotonin, 0.1, 2.5))
        self.norepinephrine = float(np.clip(self.norepinephrine, 0.1, 2.5))
        self.acetylcholine = float(np.clip(self.acetylcholine, 0.1, 2.5))

    def learning_rate_modifier(self) -> float:
        """Effective learning / plasticity gate."""
        return 0.4 * self.dopamine + 0.3 * self.acetylcholine + 0.2 * self.serotonin + 0.1

    def exploration_bonus(self) -> float:
        """Tendency to leave local maxima."""
        return 0.5 * self.dopamine + 0.3 * self.norepinephrine - 0.2 * max(0, 1.5 - self.serotonin)


# ---------------------------------------------------------------------------
# 3. Full Individual State
# ---------------------------------------------------------------------------

@dataclass
class IndividualState:
    """
    Complete state of one consciousness node (human or advanced agent).
    """
    # Core affective / performance
    happiness: float = 0.65
    productivity: float = 0.55
    energy: float = 0.75
    social_flow: float = 0.50

    # Hawkins-inspired
    hawkins_h: float = 220.0          # start slightly above courage threshold
    entropy: float = 0.45             # internal friction / resistance

    # Neuro
    neuro: NeuroState = field(default_factory=NeuroState)

    # Meta
    mode: Mode = Mode.GUARDIAN
    day: int = 0
    total_actions: int = 0
    last_feedback: float = 0.0        # -1 .. +1

    # Simple curriculum / goal tracking
    current_goal: str = "persönliche Klarheit und nachhaltige Motivation"
    subgoals_completed: int = 0
    learning_streak: int = 0

    def update_mode(self):
        self.mode = Mode.GUARDIAN if self.hawkins_h >= 200 else Mode.REACTIVE

    def entropy_from_h(self):
        """Higher H → lower entropy (MODEL). Gentle update toward target."""
        target = float(np.clip(1.0 - (self.hawkins_h / 900.0), 0.08, 0.92))
        # Move entropy only partially toward the target to avoid abrupt jumps
        self.entropy = 0.7 * self.entropy + 0.3 * target
        self.entropy = float(np.clip(self.entropy, 0.05, 0.95))

    def clip_all(self):
        self.happiness = float(np.clip(self.happiness, 0.05, 1.0))
        self.productivity = float(np.clip(self.productivity, 0.05, 1.0))
        self.energy = float(np.clip(self.energy, 0.05, 1.0))
        self.social_flow = float(np.clip(self.social_flow, 0.05, 1.0))
        self.hawkins_h = float(np.clip(self.hawkins_h, 20, 900))
        self.neuro.clip()
        self.update_mode()
        self.entropy_from_h()

    def summary(self) -> Dict:
        return {
            "day": self.day,
            "happiness": round(self.happiness, 3),
            "productivity": round(self.productivity, 3),
            "energy": round(self.energy, 3),
            "social_flow": round(self.social_flow, 3),
            "hawkins_h": round(self.hawkins_h, 1),
            "entropy": round(self.entropy, 3),
            "mode": self.mode.value,
            "dopamine": round(self.neuro.dopamine, 3),
            "serotonin": round(self.neuro.serotonin, 3),
            "learning_rate_mod": round(self.neuro.learning_rate_modifier(), 3),
            "goal": self.current_goal,
            "subgoals_completed": self.subgoals_completed,
        }


# ---------------------------------------------------------------------------
# 4. Action Effects (MODEL)
# ---------------------------------------------------------------------------

ACTION_EFFECTS: Dict[ActionType, Dict] = {
    ActionType.DEEP_WORK: {
        "happiness": 0.02,
        "productivity": 0.08,
        "energy": -0.12,
        "social_flow": -0.03,
        "hawkins": 4.0,
        "dopamine": 0.08,
        "serotonin": 0.02,
        "norepinephrine": 0.05,
        "acetylcholine": 0.10,
        "description": "Fokussierte, tiefe Arbeit an einem sinnvollen Teilziel"
    },
    ActionType.SOCIAL_SYNC: {
        "happiness": 0.07,
        "productivity": 0.02,
        "energy": -0.05,
        "social_flow": 0.12,
        "hawkins": 6.0,
        "dopamine": 0.06,
        "serotonin": 0.08,
        "norepinephrine": -0.02,
        "acetylcholine": 0.03,
        "description": "Echte Verbindung / Synchronisation mit anderen"
    },
    ActionType.REGENERATION: {
        "happiness": 0.05,
        "productivity": -0.02,
        "energy": 0.18,
        "social_flow": 0.02,
        "hawkins": 3.0,
        "dopamine": 0.03,
        "serotonin": 0.10,
        "norepinephrine": -0.08,
        "acetylcholine": 0.02,
        "description": "Schlaf, Natur, Bewegung, echte Erholung"
    },
    ActionType.CURRICULUM_REVIEW: {
        "happiness": 0.03,
        "productivity": 0.05,
        "energy": -0.04,
        "social_flow": 0.01,
        "hawkins": 5.0,
        "dopamine": 0.05,
        "serotonin": 0.04,
        "norepinephrine": 0.02,
        "acetylcholine": 0.08,
        "description": "Lernreihenfolge und Teilziele bewusst ordnen"
    },
    ActionType.SMALL_WIN: {
        "happiness": 0.09,
        "productivity": 0.04,
        "energy": 0.02,
        "social_flow": 0.03,
        "hawkins": 7.0,
        "dopamine": 0.15,
        "serotonin": 0.05,
        "norepinephrine": 0.03,
        "acetylcholine": 0.04,
        "description": "Kleinen, sichtbaren Erfolg abschließen und würdigen"
    },
    ActionType.REFLECTION: {
        "happiness": 0.04,
        "productivity": 0.01,
        "energy": -0.02,
        "social_flow": 0.00,
        "hawkins": 8.0,
        "dopamine": 0.02,
        "serotonin": 0.07,
        "norepinephrine": -0.03,
        "acetylcholine": 0.06,
        "description": "Ehrliche Bestandsaufnahme + Guardian-Loopback"
    },
    ActionType.PHYSICAL_MOVE: {
        "happiness": 0.06,
        "productivity": 0.03,
        "energy": 0.05,
        "social_flow": 0.01,
        "hawkins": 4.0,
        "dopamine": 0.07,
        "serotonin": 0.06,
        "norepinephrine": 0.04,
        "acetylcholine": 0.03,
        "description": "Körperliche Bewegung / Spaziergang"
    },
    ActionType.LEARNING_SESSION: {
        "happiness": 0.04,
        "productivity": 0.07,
        "energy": -0.08,
        "social_flow": 0.00,
        "hawkins": 5.0,
        "dopamine": 0.09,
        "serotonin": 0.03,
        "norepinephrine": 0.04,
        "acetylcholine": 0.12,
        "description": "Bewusste Lernsession mit guter Didaktik"
    },
}


# ---------------------------------------------------------------------------
# 5. Simulator Engine
# ---------------------------------------------------------------------------

class ConsciousnessSimulator:
    def __init__(self, initial: Optional[IndividualState] = None):
        self.state = initial or IndividualState()
        self.history: List[Dict] = []
        self.action_log: List[Tuple[int, str, float]] = []  # day, action, feedback
        self._record()

    def _record(self):
        self.history.append(self.state.summary())

    def apply_action(self, action: ActionType, intensity: float = 1.0, feedback: float = 0.0):
        """
        Apply one action.
        intensity : 0.5 .. 1.5 (how fully the action is performed)
        feedback  : -1 .. +1 (user-reported quality / resonance of the action)
        """
        eff = ACTION_EFFECTS[action]
        lr = self.state.neuro.learning_rate_modifier()
        intensity = float(np.clip(intensity, 0.4, 1.6))
        feedback = float(np.clip(feedback, -1.0, 1.0))

        # Base effects modulated by current neuro state and feedback
        mod = intensity * (0.7 + 0.3 * lr) * (1.0 + 0.25 * feedback)

        self.state.happiness += eff["happiness"] * mod
        self.state.productivity += eff["productivity"] * mod
        self.state.energy += eff["energy"] * mod
        self.state.social_flow += eff["social_flow"] * mod
        self.state.hawkins_h += eff["hawkins"] * mod * (1.0 - 0.3 * self.state.entropy)

        self.state.neuro.dopamine += eff["dopamine"] * mod
        self.state.neuro.serotonin += eff["serotonin"] * mod
        self.state.neuro.norepinephrine += eff["norepinephrine"] * mod
        self.state.neuro.acetylcholine += eff["acetylcholine"] * mod

        # Simple homeostasis / adaptation (MODEL)
        self.state.neuro.dopamine *= 0.97
        self.state.neuro.serotonin *= 0.98
        self.state.neuro.norepinephrine *= 0.96
        self.state.neuro.acetylcholine *= 0.97

        # Energy recovery baseline
        self.state.energy += 0.03

        # Curriculum / streak logic
        if action in (ActionType.DEEP_WORK, ActionType.LEARNING_SESSION, ActionType.CURRICULUM_REVIEW, ActionType.SMALL_WIN):
            self.state.learning_streak += 1
            if self.state.learning_streak % 3 == 0:
                self.state.subgoals_completed += 1
                self.state.hawkins_h += 3.0
        else:
            self.state.learning_streak = max(0, self.state.learning_streak - 1)

        self.state.total_actions += 1
        self.state.last_feedback = feedback
        self.state.day += 1
        self.state.clip_all()
        self.action_log.append((self.state.day, action.value, feedback))
        self._record()

    def recommend_actions(self, top_k: int = 4) -> List[Tuple[ActionType, float, str]]:
        """
        Simple ranking of actions based on current state deficits.
        Returns list of (action, score, reason).
        """
        s = self.state
        scores = []

        for act, eff in ACTION_EFFECTS.items():
            score = 0.0
            reasons = []

            # Energy management
            if s.energy < 0.45 and eff["energy"] > 0:
                score += 2.5
                reasons.append("Energie niedrig")
            if s.energy > 0.75 and eff["energy"] < 0:
                score += 0.8

            # Happiness / dopamine
            if s.happiness < 0.55 or s.neuro.dopamine < 0.9:
                score += 1.8 * max(0, eff["happiness"]) + 1.2 * max(0, eff["dopamine"])
                reasons.append("Motivation/Glück")

            # Productivity & focus
            if s.productivity < 0.55 or s.neuro.acetylcholine < 0.95:
                score += 1.5 * max(0, eff["productivity"]) + 1.0 * max(0, eff["acetylcholine"])
                reasons.append("Fokus/Produktivität")

            # Social
            if s.social_flow < 0.45:
                score += 1.6 * max(0, eff["social_flow"])
                reasons.append("soziale Verbindung")

            # Hawkins / entropy
            if s.hawkins_h < 250 or s.entropy > 0.55:
                score += 0.08 * eff["hawkins"]
                reasons.append("Bewusstseins-Frequenz / Entropie")

            # Mode-specific
            if s.mode == Mode.REACTIVE and act in (ActionType.REFLECTION, ActionType.SMALL_WIN, ActionType.REGENERATION):
                score += 1.5
                reasons.append("Guardian-Stabilisierung")

            # Learning streak encouragement
            if s.learning_streak >= 2 and act in (ActionType.DEEP_WORK, ActionType.LEARNING_SESSION, ActionType.SMALL_WIN):
                score += 0.7

            reason_str = ", ".join(reasons) if reasons else "allgemeine Balance"
            scores.append((act, score, reason_str))

        scores.sort(key=lambda x: x[1], reverse=True)
        return scores[:top_k]

    def forecast(self, action: ActionType, steps: int = 5, intensity: float = 1.0) -> List[Dict]:
        """Simple forward simulation of one action repeated (or mixed later)."""
        # Work on a copy
        import copy
        sim = ConsciousnessSimulator(copy.deepcopy(self.state))
        results = []
        for _ in range(steps):
            sim.apply_action(action, intensity=intensity, feedback=0.3)
            results.append(sim.state.summary())
        return results

    def print_status(self):
        s = self.state.summary()
        print("\n" + "="*60)
        print(f"TAG {s['day']}  |  Mode: {s['mode'].upper()}  |  H ≈ {s['hawkins_h']}")
        print("-"*60)
        print(f"Glück          : {s['happiness']:.2f}")
        print(f"Produktivität  : {s['productivity']:.2f}")
        print(f"Energie        : {s['energy']:.2f}")
        print(f"Sozialer Flow  : {s['social_flow']:.2f}")
        print(f"Entropie       : {s['entropy']:.2f}")
        print(f"Dopamin        : {s['dopamine']:.2f}   Serotonin: {s['serotonin']:.2f}")
        print(f"Lernraten-Mod  : {s['learning_rate_mod']:.2f}")
        print(f"Aktuelles Ziel : {s['goal']}")
        print(f"Teilziele ok   : {s['subgoals_completed']}   Streak: {self.state.learning_streak}")
        print("="*60)

    def print_recommendations(self):
        recs = self.recommend_actions()
        print("\nEmpfohlene nächste Handlungen (Didactic Loop):")
        for i, (act, score, reason) in enumerate(recs, 1):
            desc = ACTION_EFFECTS[act]["description"]
            print(f"  {i}. {act.value:20s}  Score {score:.2f}  – {desc}")
            print(f"      Grund: {reason}")

    def run_demo(self, days: int = 14):
        """Example autonomous run with mixed recommended actions."""
        print("Starte Demo-Simulation über", days, "Tage ...")
        for d in range(days):
            recs = self.recommend_actions(top_k=2)
            # Choose top recommendation with some randomness
            chosen = recs[0][0] if np.random.rand() < 0.7 else recs[1][0]
            # Simulate moderate positive feedback most of the time
            fb = float(np.clip(np.random.normal(0.35, 0.25), -0.3, 0.9))
            self.apply_action(chosen, intensity=1.0, feedback=fb)
            if (d + 1) % 5 == 0 or d == days - 1:
                self.print_status()

        if HAS_MPL:
            self.plot_history()

    def plot_history(self):
        if not HAS_MPL or len(self.history) < 2:
            return
        days = [h["day"] for h in self.history]
        fig, axs = plt.subplots(2, 2, figsize=(12, 8))
        axs[0, 0].plot(days, [h["happiness"] for h in self.history], "g-")
        axs[0, 0].set_title("Happiness")
        axs[0, 1].plot(days, [h["productivity"] for h in self.history], "b-")
        axs[0, 1].set_title("Productivity")
        axs[1, 0].plot(days, [h["hawkins_h"] for h in self.history], "m-")
        axs[1, 0].set_title("Hawkins H (approx)")
        axs[1, 0].axhline(200, color="gray", linestyle="--", alpha=0.7, label="Courage threshold")
        axs[1, 0].legend()
        axs[1, 1].plot(days, [h["energy"] for h in self.history], "c-")
        axs[1, 1].plot(days, [h["entropy"] for h in self.history], "r--", alpha=0.7)
        axs[1, 1].set_title("Energy (solid) / Entropy (dashed)")
        plt.tight_layout()
        plt.savefig("consciousness_sim_v011_history.png", dpi=120)
        print("\nPlot gespeichert: consciousness_sim_v011_history.png")

    def export_state(self, path: str = "consciousness_state_v011.json"):
        data = {
            "exported": datetime.now().isoformat(),
            "state": self.state.summary(),
            "history_tail": self.history[-10:],
            "action_log_tail": self.action_log[-15:],
            "status": "MODEL / GAME – didactic tool, not clinical or scientific claim"
        }
        with open(path, "w", encoding="utf-8") as f:
            json.dump(data, f, indent=2, ensure_ascii=False)
        print(f"Zustand exportiert nach: {path}")


# ---------------------------------------------------------------------------
# 6. Interactive / Demo Entry Point
# ---------------------------------------------------------------------------

def main():
    print("Collective Consciousness Simulator v0.11 – Cosmos Forge")
    print("Status: MODEL / GAME  |  Didactic & motivational tool\n")

    sim = ConsciousnessSimulator()

    # Show initial state + recommendations
    sim.print_status()
    sim.print_recommendations()

    print("\n--- Kurze Demo-Simulation (14 Tage, automatische Handlungsauswahl) ---")
    sim.run_demo(days=14)

    sim.export_state()

    print("\nFertig. Du kannst das Skript erweitern, eigene Aktionen hinzufügen")
    print("oder die recommend_actions / apply_action Methoden interaktiv nutzen.")
    print("Let’s keep refining the loop. Let’s play. Let’s be.")


if __name__ == "__main__":
    main()
```
